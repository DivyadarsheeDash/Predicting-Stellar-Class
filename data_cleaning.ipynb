```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

def trans(df): 
    col = df.columns
    cat_fea = []
    for i in col:
        if df[i].dtype == "object":
            cat_fea.append(i)
    one = OneHotEncoder()
    transformer = ColumnTransformer([("one_hot", one, cat_fea)],remainder="passthrough")
    return transformer.fit_transform(df)
```
